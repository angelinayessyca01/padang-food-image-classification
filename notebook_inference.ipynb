# 1. Introduction

Model Inference ini digunakan untuk melakukan inference menggunakan model terbaik yaitu VGG16. Model ini digunakan untuk memprediksi 9 jenis Masakan Padang menggunakan gambar baru yang tidak termasuk dalam data train, validation atau test.

Tujuan inference adalah untuk melihat kemampuan model dalam mengenali gambar baru setelah proses traning dan evaluasi selesai dilakukan.

# 2. Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.applications.vgg16 import preprocess_input
from tensorflow.keras.utils import load_img, img_to_array
from google.colab import files
```

# 3. Load Model

```python
# Upload Best Model
best_model = tf.keras.models.load_model('best_model_vgg16.keras')
```

# 4. Nama Kelas di Dataset

```python
# Nama Kelas di Dataset
nama_kelas = [
    'ayam_goreng',
    'ayam_pop',
    'daging_rendang',
    'dendeng_batokok',
    'gulai_ikan',
    'gulai_tambusu',
    'gulai_tunjang',
    'telur_balado',
    'telur_dadar'
]

print('Jumlah Kelas :', len(nama_kelas))
print(nama_kelas)
```

# 5. Preprocessing + Prediction

## 5.1 Prediksi Gambar Ayam Goreng

```python
# Image path
img_path = '/content/images/ayam_goreng.jpg'

# Load dan resize gambar
img = load_img(
    img_path,
    target_size = (224, 224)
)

# Convert image to array
img_array = img_to_array(img)

# Tambahkan dimensi batch
img_array = np.expand_dims(img_array, axis=0)

# VGG16 preprocessing
img_array = preprocess_input(img_array)

# Prediction
prediction = best_model.predict(img_array)

# Get predicted class
predicted_index = np.argmax(prediction)
predicted_class = nama_kelas[predicted_index]
confidence = np.max(prediction)

print('Predicted Class :', predicted_class)
print(f'Confidence      : {confidence * 100:.2f}%')
```
